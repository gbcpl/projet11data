```python
df_final = dispo_poulet.copy()

df_final = df_final.merge(population_2017, on="Country", how="left")
df_final = df_final.merge(corruption_2017, on="Country", how="left")
df_final = df_final.merge(stability, on="Country", how="left")
df_final = df_final.merge(gov_effectiveness, on="Country", how="left")
df_final = df_final.merge(growth, on="Country", how="left")
df_final = df_final.merge(inflation, on="Country", how="left")
df_final = df_final.merge(gdp_per_capita, on="Country", how="left")
df_final = df_final.merge(infrastructures, on="Country", how="left")
df_final = df_final.merge(ges_per_capita, on="Country", how="left")
df_final = df_final.merge(renewables, on="Country", how="left")
df_final = df_final.merge(tariffs, on="Country", how="left")
```
